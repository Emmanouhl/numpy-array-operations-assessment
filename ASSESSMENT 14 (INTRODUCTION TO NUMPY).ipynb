```python
# Import NumPy for the bonus challenge
import numpy as np
```

```python
# Seven days of recorded temperatures, converted to a NumPy array
temperatures = [27.5, 29.0, 30.5, 28.0, 31.0, 32.5, 30.0]
temperatures_array = np.array(temperatures)
print(temperatures_array)
```

```python
#The Temperature array is 1D
#It is a Vector

# .dtype shows what type of numbers the array holds (here, floating point)
print(temperatures_array.dtype)
#The data type of the array is "float"

# len() gives the number of elements in the array (i.e. number of readings)
print(len(temperatures_array))
#The Temperature array has 7 observations
```

```python
# Target temperatures for the same seven days, also as a NumPy array
target_temperature = [28.0, 28.0, 29.0, 28.0, 30.0, 30.0, 29.0]
target_temperature_array = np.array(target_temperature)
```

```python
# Element-wise subtraction: how far each day's recorded temperature is
# from the target (positive = above target, negative = below target)
temperature_difference = temperatures_array - target_temperature_array
print(f"Temperature Difference: {temperature_difference}")
```

```python
# Add 2 degrees to every recorded temperature (broadcasting: the scalar 2
# is applied to each element of the array automatically)
increase_recorded_temperature = (temperatures_array) + 2
print(f"Increased Recorded Temperature: {increase_recorded_temperature}")

# Same broadcasting applied to the target temperatures
increase_target_temperature = (target_temperature_array) + 2
print(f"Increased Target Temperature: {increase_target_temperature}")
```

```python
# Import NumPy for the bonus challenge
import numpy as np
```

```python
#This represent the temperature of three locations from Monday - Friday
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.0]

# Convert each location's temperature list into a NumPy array
lagos_array = np.array(lagos)
abuja_array = np.array(abuja)
ibadan_array = np.array(ibadan)

print(f"Lagos Temperature: {lagos_array}")
print(f"Abuja Temperature: {abuja_array}")
print(f"Ibadan Temperature: {ibadan_array}")
```

**Challenge 1: Which location recorded the highest tempeature on each day?**

    Monday = Lagos
    Tuesday = Lagos
    Wednesday = Lagos
    Thursday = Abuja
    Friday = Abuja

**Challenge 2: Create a calculation that increases every recorded temperature by 1°C.**

```python
#Increase in every recorded temperature by 1 degree C
# Broadcasting adds 1 to every element of each array in one operation
increase_lagos_array = (lagos_array) + 1
increase_abuja_array = (abuja_array) + 1
increase_ibadan_array =(ibadan_array) + 1

print(f"Increased Lagos Temperature: {increase_lagos_array}")
print(f"Increased Abuja Temperature: {increase_abuja_array}")
print(f"Increased Ibadan Temperature: {increase_ibadan_array}")
```

**Challenge 3: Create a calculation that compares Lagos temperatures with Abuja temperatures.**

```python
#Temperature difference between lagos and Abuja locations
# Positive values = Lagos warmer that day; negative values = Abuja warmer
temperature_difference = lagos_array - abuja_array
print(f"Lagos and Abuja Temperature Comparison: {temperature_difference}")
```

**Challenge 4: Explain why NumPy becomes increasingly useful as the amount of numerical data grows.**

    NumPy becomes increasingly useful for large datasets because of its performance and efficiency. This means mathematical operations on large arrays happen much faster and use significantly less memory. Additionally, it eliminates the need for slow Python for loops by allowing operations to be applied to the entire array at once (element-wise operations).
